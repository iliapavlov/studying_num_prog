```python
import numpy as np
import matplotlib.pyplot as plt


x1 = 1
x2 = 3
y1 = 2
y2 = 6


slope = (y2-y1) / (x2-x1)


m = slope
b = y1 - slope * x1


x_range = np.linspace(x1, x2, num=11)


y_range = m*x_range + b


plt.plot(x_range, y_range, color="red")
plt.scatter([x1, x2], [y1, y2], color="blue")


plt.show()
```

```python
import pandas as pd
df = pd.DataFrame({'Time':[0,20,40,60,80,100], 'Temp':[26.0, 48.6, 61.6, 71.2, 74.8, 75.2]})
df
```

```python
# Відображення результатів
plt.figure(figsize=(8, 4))
plt.plot(df['Time'], df['Temp'], 'o', label='Вихідні дані')
plt.plot(df['Time'], df['Temp'], '-', label='Лінійна інтерполяція')


# Визначення точок спостережень, між якими буде проведена лінія
observation_points = [50]


# Додавання вертикальної пунктирної лінії між вказаними точками
for date in observation_points:
    plt.vlines(date, ymin=df['Temp'].min(), ymax=df['Temp'].max(), linestyle='dashed', color='gray')


plt.title('Приклад лінійної інтерполяції')
plt.xlabel('Час')
plt.ylabel('Температура')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import splev, splrep


# Задаємо дані для інтерполяції
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 3, 1, 4, 7])


# Побудова кубічного сплайна
spl = splrep(x, y, k=3)


# Генерація нових точок для візуалізації сплайна
x_new = np.linspace(min(x), max(x), 1000)
y_new = splev(x_new, spl)


# Візуалізація результатів
plt.scatter(x, y, label='Дані')
plt.plot(x_new, y_new, label='Кубічний сплайн', color='red')
plt.title('Інтерполяція кубічним сплайном')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import Rbf


# Дані
points = np.array([[0, 1], [1, 4], [2, 2], [3, 8], [4, 6]])


# Виділення окремо x та y координат
x = points[:, 0]
y = points[:, 1]


# Параметр ширини функції
sigma = 1.0


# Створення об'єкта Rbf для інтерполяції
rbf = Rbf(x, y, function='gaussian', epsilon=sigma)


# Проміжні точки для інтерполяції
interpolation_points = np.linspace(0, 4, 100)
interpolated_values = rbf(interpolation_points)


# Візуалізація результатів
plt.scatter(x, y, label='Дані')
plt.plot(interpolation_points, interpolated_values, label='Інтерполяція (Гаусівська РБФ)', color='red')
plt.title('Інтерполяція гаусівською РБФ за допомогою SciPy')
plt.legend()
plt.show()
```
